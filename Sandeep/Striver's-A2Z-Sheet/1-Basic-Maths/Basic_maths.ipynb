### 1. Count the digits

#### 1. Brute Force method!

```python
def count_digits(number: int) -> int:
    sign = 1
    if number < 0:
        number = abs(number)

    ## We integer divide and add the result into a variable
    counter = 0
    while number > 0:
        counter += 1 
        number = number // 10

    if counter == 0:
        print("This is 0, being the only digit")
    
    return counter

## Runner
N = int(input("Enter your desired number:\n"))
print(f"You have inputed: {N}")
print(count_digits(N))
```

#### 2. Optimal method!

```python
import math
N = 0
## We would use log base 10 for the answer
def count_digits_better(number: int) -> int:
    counter = 0
    power = math.log(abs(number), 10)
    counter = power + 1
    return int(counter) 

## Runner
N = int(input("Enter your desired number:\n"))
print(f"You have inputed: {N}")
print(count_digits_better(N))
```

### 2. Reverse the number
- No trailing zeroes by the leading ones

```python
def reverse_given_number(number: int) -> int:
    ## handle negative numbers
    sign = 1 
    if number < 0:
        sign = -1
        number = abs(number)

    ## We would use maths for this!
    reversed_number = 0

    while number > 0:
        ## Gets the last digit
        last_digit = number % 10
        ## constructs the number back from the recently pop one
        reversed_number = (reversed_number * 10) + last_digit
        ## decrease the digit
        number = number // 10
    
    return reversed_number * sign

## Runner
N = int(input("Enter your desired number:\n"))
print(f"You have inputed: {N}")
print(reverse_given_number(N))
```

### 3. Is the number palindrome?

```python
def is_palindrome(number: int) -> bool:
    # if it's palindrome, then the reverse must be the same as the number itself
    return reverse_given_number(number) == number

## Runner
N = int(input("Enter your desired number:\n"))
print(f"You have inputed: {N}")
print(f"This is the number palindrome?: {is_palindrome(N)}")
```

### 4. GCD of two numbers


```python
## as we get the factors we need to modify so that
## we move for the one with the minimum of the 2, and try to find the common one 
def GCD(number1: int, number2: int) -> int:
    ## we rotate through the smaller one, 
    ## and check if the number divides both, 
    ## if yes, change the main one to it, else move on
    ## and start moving backwards from the smallest one, 
    ## as soon as u encounter the highest one, and return 
    gcf = 0
    iterate_using = min(number1, number2)
    for number in range(iterate_using, 0, -1):
        if number1 % number == 0 and number2 % number == 0 and gcf < number:
            gcf = number
            return gcf
            
            break

GCD(77, 88)
```

### 5. Is it an armstrong number?
- An Amrstrong number is a number that is equal to the sum of its own digits each raised to the power of the number of digits. 

```python
def is_armstrong(number: int, digits: int) -> bool:
    if number < 0:
        number = abs(number)
    armstrong = 0
    ## we copy the number instance into another variable
    copy_number = number

    while copy_number > 0:
        ## Gets the last digit
        last_digit = copy_number % 10

        ## constructs the number back from the recently pop one
        armstrong += last_digit ** digits

        ## decrease the digit
        copy_number = copy_number // 10
    
    return armstrong == number
    
N = 153
is_armstrong(N, count_digits_better(N))
```

### 6. Print all Divisors of a given Number

```python
def all_factors(number: int) -> list[int]:
    factors = []
    for num in range(1, int(number**0.5) + 1):
        if number % num == 0:
            factors.append(num)
            if num != number // num:
                factors.append(number // num)
    return factors

print(all_factors(36))
```

### 7. Is the number prime?

```python
N = 23
if len(all_factors(N)) == 2:
    print("It's a prime number!")
else:
    print("It's not a prime number!")
```
